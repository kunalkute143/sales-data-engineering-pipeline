```python
import pandas as pd

kk = pd.read_csv('C:/online_retail.csv')
kk.head()
```

```python
kk.info()
```

```python
kk.isnull().sum()
```

```python

kk_clean = kk.dropna(subset=['CustomerID'])

kk_clean
```

```python
kk_clean['InvoiceDate'] = pd.to_datetime(kk_clean['InvoiceDate'])
kk_clean['InvoiceDate']
```

```python
kk_clean['Revenue'] = kk_clean['Quantity'] * kk_clean['UnitPrice']
kk_clean.head()
```

```python
from sqlalchemy import create_engine

engine = create_engine('mysql+mysqlconnector://root:Kunal%40143@localhost/sales_warehouse')

kk_clean.to_sql('sales', engine, if_exists='replace', index=False)

print("Data loaded to MySQL successfully!")
```

```python
from pyspark.sql import SparkSession


spark = SparkSession.builder.appName("SalesPipeline").getOrCreate()


spark_df = spark.createDataFrame(kk_clean)

spark_df.show(5)
```

```python
from pyspark.sql.functions import sum as spark_sum

# Country नुसार total revenue काढ (PySpark मध्ये)
spark_df.groupBy("Country").agg(spark_sum(spark_df.Quantity * spark_df.UnitPrice).alias("TotalRevenue")) \
    .orderBy("TotalRevenue", ascending=False) \
    .show(10)
```
